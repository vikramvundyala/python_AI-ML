```python
from sklearn.datasets import load_diabetes
import pandas as pd

X, y = load_diabetes(return_X_y=True)

X.shape
```

```python
y.shape
```

```python
df_X = pd.DataFrame(data=X, columns=["age","sex","bmi","bp", "tc", "ldl", "hdl","tch", "ltg", "glu"])
df_X.head()
```

```python
df_y=pd.DataFrame(data=y, columns=["disease_progression"])
df_y.head()
```

```python
df = pd.merge(df_X, df_y, left_index=True, right_index=True)
df.head()
```

```python
df.info()
```

```python
df.describe()
```
