```
# YOUR CODE HERE for downloading and splitting the multiple stories of respective authors which are allocated to you

from urllib import request
from nltk.tokenize import sent_tokenize
import re
import pandas as pd

url = "https://www.gutenberg.org/files/2781/2781-0.txt"
response = request.urlopen(url)
raw_A = response.read().decode('utf8')

text_split_A = re.split(r'[\r\n\r\n\r\n\r\n\r\n]', raw_A)
len(text_split_A)
df = pd.DataFrame({'Paragraphs': [para.strip() for para in text_split_A if para.strip() != '']})
stories_list_A = df.iloc[17:29]
```

```
stories_list_A
```

```
story_Array = []
i =29;
for outer_idx, outer_row in df.iloc[17:29].iterrows():
    story_chunk = []
    for inner_idx, inner_row in df.iloc[i:].iterrows():
        if outer_row["Paragraphs"].strip() == inner_row["Paragraphs"].strip():
            break
        story_chunk.append(inner_row["Paragraphs"])
        i += 1
    if(len(story_chunk)>0):
     story_Array.append(story_chunk)

df_A = pd.DataFrame({'Story': [s for s in story_Array if s != '']})
len(story_Array)

```

### 1 Marks ->  a) List 6 handcrafted features to distinguish author stories.

```
# For eg:
# 1. UniqueWords
# 2. AvgSentLength
# List the other handcrafted features here

from nltk.tokenize import sent_tokenize,word_tokenize
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words('english'))
punctuation = set(string.punctuation)

def clean_text(text: str) :
  words = word_tokenize(text)
  words = [word.lower() for word in words if word.isalpha()]
  words = [word for word in words if word not in stop_words]
  words = [word for word in words if word not in punctuation]
  return words


df_All['sentence_length'] = df_All['Story'].apply(lambda x: len(sent_tokenize(x)))
df_All['clean_text'] = df_All['Story'].apply(lambda x: clean_text(x))
df_All['unique_word_length'] = df_All['clean_text'].apply(lambda x: len(word_tokenize(x)))

avg = df_All["unique_word_length"].sum() / len(df_All)
df_All['avg_word_length'] = avg

# YOUR CODE HERE



```
